## Key takeaways

1. **1,000,000 transactions** with **7 features** and **1 target** (`fraud`).
2. **No missing values and no duplicates**, so no rows had to be removed.
3. All columns are stored as `float64`, including the **binary features** (`repeat_retailer`, `used_chip`, `used_pin_number`, `online_order`) and the target, which only contain 0.0 and 1.0. They could be converted to integers.
4. **The data is imbalanced:** 87,403 fraud cases (**8.74%**) vs. 912,597 non-fraud (**91.26%**).
   - A model that always predicts "not fraud" would already reach **91.26% accuracy** while catching zero fraud.
   - So accuracy alone is misleading here: the model needs to be judged on **recall** and **precision**, and the train-test split should be **stratified**.
5. **The three continuous features are strongly right-skewed** (mean far above the median, with extreme maximums):

   | Feature | Median | Mean | Max |
   |---|---|---|---|
   | distance_from_home | 9.97 | 26.63 | 10,632.72 |
   | distance_from_last_transaction | 1.00 | 5.04 | 11,851.10 |
   | ratio_to_median_purchase_price | 1.00 | 1.82 | 267.80 |

   These extreme values are **not removed**: in fraud detection, unusual transactions can be exactly the signal the model needs. But they matter for **distance-based models like KNN**, so scaling or a log transform should be tested.
6. **Binary features** (share of transactions with value 1): `repeat_retailer` 88.2%, `online_order` 65.1%, `used_chip` 35.0%, `used_pin_number` 10.1%.
